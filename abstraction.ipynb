```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("barking")
d = Dog()
d.sound()
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
class UPI(Payment):
    def pay(self,amount):
        print(f"{amount}")
class Check(Payment):
    def pay(self,amount):
        print(f"{amount}")
c = Check()
u = UPI()
c.pay(5000)
u.pay(2000)

```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    def validate_user(self):
        print("Valid user")
    @abstractmethod
    def pay(self,amount):
        pass
class UPI(Payment):
    def pay(self,amount):
        print(f"{amount}")

u = UPI()
u.validate_user()
u.pay(2000)
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    def __init__(self,amount):
        self.amount = amount

    @abstractmethod
    def pay(self,amount):
        pass
class UPI(Payment):
    def pay(self):
        print(f"{self.amount}")

u = UPI(2000)
u.pay()
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self,amount):
        pass
    @abstractmethod
    def check(self,amount):
        pass
class UPI(Payment):
    def pay(self,amount):
        print(f"{amount}")
    def check(self,amount):
        print(f"{amount}")

u = UPI()
u.pay(2000)
u.check(500)
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass
class UPI(Payment):
    def pay(self, amount):
        print("UPI Payment:", amount)
class Card(Payment):
    def pay(self, amount):
        print("Card Payment:", amount)

payments = [UPI(),Card()]
for payment in payments:
    payment.pay(2000)
```

```python
from abc import ABC,abstractmethod

class Student(ABC):
    def __init__(self,name):
        self.name = name
    def display(self):
        print("Hi",self.name)
    @abstractmethod
    def work(self):
        pass
class Child(Student):
    def work(self):
        super().display()
        print("Hi")
c = Child("Geerthy")
c.work()
```

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    @property
    @abstractmethod
    def salary(self):
        pass
class Developer(Employee):
    @property
    def salary(self):
        return 50000
d = Developer()
print(d.salary)
```

```python
class Student:

    def __init__(self):
        self.__age = 23

    def set_age(self, age):

        if age > 0:
            self.__age = age
        else:
            print("Invalid Age")

    def get_age(self):
        return self.__age

student = Student()

student.set_age(-5)

print(student.get_age())
```

```python
class BankAccount:

    def __init__(self):
        self.__balance = 1000

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient Balance")

    def get_balance(self):
        return self.__balance

account = BankAccount()

account.deposit(500)

account.withdraw(200)

print(account.get_balance())
```

```python
class Student:

    def __init__(self):
        self.__age = 23

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value > 0:
            self.__age = value
        else:
            print("Invalid Age")

student = Student()

print(student.age)

student.age = 30

print(student.age)

student.age = -5
```

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name          # Public
        self.__salary = salary    # Private

    def show_salary(self):
        return self.__salary

    def increase_salary(self, amount):
        if amount > 0:
            self.__salary += amount

employee = Employee("Geerthy", 25000)

print(employee.name)
print(employee.show_salary())

employee.increase_salary(5000)

print(employee.show_salary())
```

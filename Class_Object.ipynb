```python
class Dog:
    def __init__(self, name, food):
        self.name = name
        self.food = food
d = Dog("Tommy", "Dog Food")
print(f"Dog name is {d.name} and food is {d.food}")
```

```python
class car:
    def __init__(self,name,model,brand):
        self.name = name
        self.model = model
        self.brand = brand
    def display(self):
        print(f"Car name is {self.name} and model is {self.model} and brand is {self.brand}")
c1 = car("BMW", "X5", "BMW")
c2 = car("Audi", "A6", "Audi")
c1.display()
c2.display()
```

```python
class students:
    def __init__(self,name,age,place):
        self.name = name
        self.age = age
        self.place = place
    def student_details(self):
        print(f"Hi my name is {self.name}")
    def info(self):
        print(f"My age is {self.age} and I am from {self.place}")
s = students("Geerthy",23,"chemmanvilai")
s.student_details()
s.info()
```

```python
class students:
    school_name = "St. Joseph's Matriculation Higher Secondary School"
    def __init__(self, name, age, roll_no):
        self.name = name
        self.age = age
        self.roll_no = roll_no
    def display(self):
        print(f"Student name is {self.name} and age is {self.age} and roll number is {self.roll_no} and school is {self.school_name}")

s = students("Geerthy", 23, 101)
s.display() 
```

```python
class students:
    school_name = "St. Joseph's Matriculation Higher Secondary School"
    def __init__(self, name):
        self.name = name

    @classmethod
    def change_school(cls, new_name):
        cls.school_name = new_name  

s1 = students("Arun")
s2 = students("Divya")

students.change_school("XYZ School")
print(s1.school_name)
print(s2.school_name)  
```

```python
class math:
    @staticmethod
    def add(x,y):
        return x+y
    def multiply(x,y):
        return x*y  
print(math.add(5, 10))
print(math.multiply(5, 10))
```

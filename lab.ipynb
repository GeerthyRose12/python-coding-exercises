```python
Student_Name = input("Enter your name :")
Student_ID = int(input("Enter your student_ID :"))
Test_Score = int(input("Enter your test score :"))
if Test_Score >= 95 and Test_Score <= 100:
    grade = "A"
elif Test_Score >= 90 and Test_Score <= 94:
    grade = "B"
elif Test_Score >= 85 and Test_Score <= 89:
    grade = "C"
elif Test_Score >= 80 and Test_Score <= 84:
    grade = "D" 
elif Test_Score >= 70 and Test_Score <= 79:
    grade = "E"
else:
    grade = "F"
print("Student Name :", Student_Name)
print("Student ID :", Student_ID)
print("Test Score :", Test_Score)
print("Grade :", grade)
```

```python
Account_Holder_Name = "Geerthy Rose"
PIN = 9677
Account_Balance = 10000

ATM_PIN = int(input("Enter your ATM PIN :"))
withdrawal_Amount = int(input("Enter the amount to withdraw :"))
if ATM_PIN == PIN :
    if withdrawal_Amount <= Account_Balance:
        print(f"Welocome {Account_Holder_Name}!\nTransaction Successful")
        print(f"Your withdrawal amount is: {withdrawal_Amount}")
        Balance_Amount = Account_Balance - withdrawal_Amount
        print(f"Your remaining balance is: {Balance_Amount}")
    else:
        print("Not enough balance.")
else:
    print("Invalid PIN.")
```

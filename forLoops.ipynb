```python
No_of_Emp = int(input("Enter the number of employess: "))
total_salary = 0
for i in range(1,3+1):
    Emp_Id = int(input(f"Enter the EmployeeID {i}: "))
    Emp_name = input(f"Enter the name {i}: ")
    Emp_salary = int(input(f"Enter the salary{i}: "))
    total_salary += Emp_salary
    average = total_salary/3
    print(Emp_Id,Emp_name,Emp_salary,)
print("Number of Employees : ",No_of_Emp)
print("Totalsalary : ",total_salary)
print("Average:",average)

```

```python
emp_id_list = []
emp_names_list = []
emp_salary_list = []
No_of_Emp = int(input("Enter the number of employess: "))
total_salary = 0
for i in range(1,No_of_Emp+1):
    Emp_id = int(input(f"Enter the EmployeeID {i}: "))
    Emp_name = input(f"Enter the name {i}: ")
    Emp_salary = int(input(f"Enter the salary{i}: "))
    emp_id_list.append(Emp_id)
    emp_names_list.append(Emp_name)
    emp_salary_list.append(Emp_salary)

print("----Employee Details----")
for i in range(len(emp_id_list)):
    print("EmployeeId: ", emp_id_list[i])
    print("EmployeeName: ",emp_names_list[i])
    print("EmployeeSalary:",emp_salary_list[i])

total = sum(emp_salary_list)
print("\n ---Salary details---")
print("Total Salary:",total)
print("Average Salary:",total/No_of_Emp)
print("Highest_salary:",max(emp_salary_list))

    


```

```python
rows = int(input("Enter the number of rows: "))
seats = int(input("Enter the number of seats: "))
for i in range(rows):
    for j in range(1,seats):
        print(chr(65+i)+str(j),end=" ")
    print()
```

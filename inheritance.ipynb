```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)


class Student(Person):

    def study(self):
        print(self.name, "is studying")


student = Student("Geerthy", 23)

student.display()
student.study()
```

```python
class Duck:

    def speak(self):
        print("Quack")


class Dog:

    def speak(self):
        print("Bark")


class Robot:

    def speak(self):
        print("Beep")

def make_speak(obj):
    obj.speak()

make_speak(Duck())
make_speak(Dog())
make_speak(Robot())
```

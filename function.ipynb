```python
def addition(a,b):
    return a+b
def multiply(a,b):
    return a*b
def calculate(a,b):
    sum_result = addition(a,b)
    product_result = multiply(a,b)
    return sum_result, product_result
s,p =calculate(5,10)
print(s, p)
```

```python
def outer_function(x):
    def inner_function(y):
        return x + y
    return inner_function
print(outer_function(5)(10)) 
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)
factorial(5)
```

```python
def apply_operation(a, b, operation):
    return operation(a, b)  

def add(a, b): 
    return a + b
def subtract(a, b): 
    return a - b

print(apply_operation(10, 5, add))       
print(apply_operation(10, 5, subtract))  
print(apply_operation(10, 5, lambda x, y: x * y))
```

```python
def log_function_call(func):
    def wrapper(*args, **kwargs):
        print(f"Calling function: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Finished: {func.__name__}")
        return result
    return wrapper

@log_function_call
def add(a, b):
    return a + b

print(add(3, 4))
```

```python
import csv

with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Mark"])
    writer.writerow(["Arun", 89])
    writer.writerow(["Priya", 95])
```

```python
with open("students.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
import json

data = {"name": "Arun", "age": 22, "skills": ["Python", "SQL"]}
with open("data.json", "w") as f:
    json.dump(data, f, indent=4)
```

```python
with open("data.json", "r") as f:
    loaded = json.load(f)
    print(loaded["name"])
```

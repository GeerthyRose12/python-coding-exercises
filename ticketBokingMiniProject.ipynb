```python
No_of_rows = int(input("Enter Number of rows :"))
No_of_seats = int(input("Enter Number of seats per row: "))
seat_numbers = []

while True:
    print("====Ticket Booking System====")
    print("\t1.View seats")
    print("\t2.Book Ticket")
    print("\t3.View Booked seats")
    print("\t4.View Available Seats Count")
    print("\t5.Exit")

    choice = int(input("Enter your choice : "))
    if choice == 1:
        print("="*30)
        print("====Seats In Theatre=======")
        print("="*30)
        for row in range(No_of_rows):
            row_name = chr(65 + row)
           
            for seat in range(1, No_of_seats):
                current_seat = f"{row_name}{seat}"
                if current_seat in seat_numbers:
                    print("X" , end ="\t")
                else:
                    print(current_seat, end = "\t")
            print()
        print("="*30)
    if choice == 2:
        seat_number = input("Enter the seat Number (Example : A1):").upper()
        valid_seat = False
        for row in range(No_of_rows):
            row_name = chr(65 + row)
            for seat in range(1, No_of_seats):
                current_seat = f"{row_name}{seat}"
                if current_seat == seat_number:
                    valid_seat = True
                            
                        
                             
                    
        if not valid_seat:
            print("Its not valid seat")
        elif seat_number in seat_numbers:
            print("Already Booked")
        else:
            seat_numbers.append(seat_number)
            print("Ticket Booked Succesfully")
            print("==Updated Sheet==")
            for row in range(No_of_rows):
                row_name = chr(65 + row)
                for seat in range(1, No_of_seats):
                    current_seat = f"{row_name}{seat}"
                    if current_seat in seat_numbers:
                        print("X" , end = "\t")
                    else:
                        print(current_seat, end = "\t")
                print()
    if choice == 3:
        if len(seat_numbers) == 0:
            print("No seats are booked yet")
        else:
            for seat in seat_numbers:
                print("===Booked seats===")
                print(seat)
    if choice == 4:
        total_seats = No_of_rows * No_of_seats
        booked_seats = len(seat_numbers)
        available_seats = total_seats - booked_seats
        print("==== Available Seats Count ====")
        print("Total Seats:", total_seats)
        print("Booked Seats:", booked_seats)
        print("Available Seats:", available_seats)
    if choice == 5:
        print("Thank You")
        break











```

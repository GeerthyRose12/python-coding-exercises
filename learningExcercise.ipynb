```python
class dog:
    def Bow(self):
        print("Bow")
class Cat(dog):
    def Meow(self):
        print("Meow")


c = Cat()
c.Bow()
c.Meow()
```

```python
from abc import ABC,abstractmethod
class Teacher(ABC):
    @abstractmethod
    def details(self):
        pass

class student(Teacher):
    def details(self):
        print("ggggg")
s = student()
s.details()
```
